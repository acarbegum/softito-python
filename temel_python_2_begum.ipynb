### 🔸 Soru 7:
Bir yılın artık yıl olup olmadığını kontrol edin.
Artık yıl kuralı:
- 4'e tam bölünebiliyorsa artık yıldır
- Ama 100'e bölünebiliyorsa artık yıl **değildir**
- Ama 400'e bölünebiliyorsa yine artık yıldır

```python
# Cevabınızı buraya yazın:
# Test: 2000 (artık), 1900 (değil), 2024 (artık), 2023 (değil)

def artik_yil_mi(yil):
    return (yil % 4 == 0 and yil % 100 != 0) or (yil % 400 == 0)

print(artik_yil_mi(2000))
print(artik_yil_mi(1900))
print(artik_yil_mi(2024))
print(artik_yil_mi(2023))
```

### 🔸 Soru 12:
Bir öğrenci listesi var. Her öğrenci `(isim, not)` tuple'ı olarak saklanıyor.
- Notu 70'in üzerinde olanları filtreleyin
- En yüksek notu alan öğrenciyi bulun
- Notların ortalamasını hesaplayın

```python
# Cevabınızı buraya yazın:
ogrenciler = [
    ("Ahmet", 85),
    ("Büşra", 62),
    ("Can", 91),
    ("Deniz", 74),
    ("Elif", 58),
    ("Furkan", 88)
]

basarili_ogrenciler = [isim for isim, notu in ogrenciler if notu > 70]

en_iyi_ogrenci = max(ogrenciler, key=lambda x: x[1])

ortalama = sum(notu for _, notu in ogrenciler) / len(ogrenciler)

print("70 üstü öğrenciler:", basarili_ogrenciler)
print("En yüksek not:", en_iyi_ogrenci)
print("Ortalama:", ortalama)
```

### 🔸 Soru 13 (Set):
İki sınıfın öğrenci listesi var. `set` kullanarak:
- Her iki sınıfta da bulunan öğrencileri bulun (kesişim)
- Toplam kaç farklı öğrenci olduğunu bulun (birleşim)
- Sadece A sınıfında olup B'de olmayan öğrencileri bulun (fark)

```python
# Cevabınızı buraya yazın:
sinif_A = {"Ali", "Ayşe", "Can", "Dilan", "Ece"}
sinif_B = {"Can", "Dilan", "Fatih", "Gökçe", "Hakan"}

kesisim = sinif_A & sinif_B

birlesim = sinif_A | sinif_B

fark = sinif_A - sinif_B

print("Kesişim:", kesisim)
print("Toplam farklı öğrenci:", len(birlesim))
print("A'da olup B'de olmayanlar:", fark)
```

### 🔸 Soru 14:

### 🔸 Soru 15:
Özyinelemeli (recursive) bir fonksiyon yazın: `faktoriyel(n)` - n! hesaplasın.
- 0! = 1
- n! = n × (n-1)!

5!, 10! değerlerini hesaplayın.

```python
# Cevabınızı buraya yazın:
def faktoriyel(n):
    if n == 0:
        return 1
    return n * faktoriyel(n - 1)

print(faktoriyel(5))
print(faktoriyel(10))
```

### 🔸 Soru 16:
`lambda` fonksiyonu kullanarak:
- Bir sayının karesini hesaplayan lambda

### 🔸 Soru 18 (Bütünleştirici):
Basit bir **öğrenci not sistemi** yazın:

- `ogrenciler` adında boş bir sözlük oluşturun (anahtar: isim, değer: notlar listesi)
- `ogrenci_ekle(isim)`: Yeni öğrenci ekler
- `not_ekle(isim, not_)`: Öğrenciye not ekler. Öğrenci yoksa hata mesajı verir.
- `ortalama_hesapla(isim)`: Öğrencinin not ortalamasını döndürür
- `sinif_raporu()`: Tüm öğrencileri ve ortalamalarını yazdırır, sınıf ortalamasını gösterir

En az 3 öğrenci ve her birine 3 not girerek test edin.

```python
# Cevabınızı buraya yazın:
ogrenciler = {}

def ogrenci_ekle(isim):
    if isim not in ogrenciler:
        ogrenciler[isim] = []
    else:
        print("Öğrenci zaten var!")

def not_ekle(isim, not_):
    if isim in ogrenciler:
        ogrenciler[isim].append(not_)
    else:
        print("Hata: Öğrenci bulunamadı!")

def ortalama_hesapla(isim):
    if isim in ogrenciler and len(ogrenciler[isim]) > 0:
        return sum(ogrenciler[isim]) / len(ogrenciler[isim])
    return 0

def sinif_raporu():
    toplam_ortalama = 0

    for isim in ogrenciler:
        ort = ortalama_hesapla(isim)
        print(f"{isim}: {ort}")
        toplam_ortalama += ort

    if len(ogrenciler) > 0:
        sinif_ort = toplam_ortalama / len(ogrenciler)
        print("Sınıf ortalaması:", sinif_ort)

# Test kodu:
ogrenci_ekle("Ahmet")
ogrenci_ekle("Büşra")
ogrenci_ekle("Can")

not_ekle("Ahmet", 85)
not_ekle("Ahmet", 90)
not_ekle("Ahmet", 78)

not_ekle("Büşra", 92)
not_ekle("Büşra", 88)
not_ekle("Büşra", 95)

not_ekle("Can", 70)
not_ekle("Can", 65)
not_ekle("Can", 73)

sinif_raporu()
```
